```python

data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'No'],
    ['Sunny', 'Warm', 'High', 'Weak', 'Yes']
]

h = ['0'] * 4

for row in data:
    if row[-1] == 'Yes':
        for i in range(4):
            if h[i] == '0':
                h[i] = row[i]
            elif h[i] != row[i]:
                h[i] = '?'

print("Final Hypothesis:", h)
```

```python

import pandas as pd
from sklearn.tree import DecisionTreeClassifier, export_text

data = pd.DataFrame({
    'Outlook': ['Sunny','Sunny','Overcast','Rain','Rain','Rain','Overcast','Sunny'],
    'Humidity': ['High','High','High','High','Normal','Normal','Normal','High'],
    'Wind': ['Weak','Strong','Weak','Weak','Weak','Strong','Strong','Weak'],
    'Play': ['No','No','Yes','Yes','Yes','No','Yes','No']
})

X = pd.get_dummies(data.drop('Play', axis=1))
model = DecisionTreeClassifier(criterion='entropy', random_state=0)
model.fit(X, data['Play'])

print(export_text(model, feature_names=list(X.columns)))

sample = pd.DataFrame([{'Outlook':'Sunny','Humidity':'Normal','Wind':'Weak'}])
sample = pd.get_dummies(sample).reindex(columns=X.columns, fill_value=0)
print("Prediction:", model.predict(sample)[0])
```

```python

import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_squared_error

X = np.array([[1],[2],[3],[4],[5],[6]])
y = np.array([1,4,9,16,25,36])

linear = LinearRegression().fit(X, y)
poly = make_pipeline(PolynomialFeatures(2), LinearRegression()).fit(X, y)

print("Linear MSE:", mean_squared_error(y, linear.predict(X)))
print("Polynomial MSE:", mean_squared_error(y, poly.predict(X)))
print("Linear Predictions:", linear.predict(X))
print("Polynomial Predictions:", poly.predict(X))
```

```python

import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import make_pipeline
from sklearn.ensemble import RandomForestRegressor

data = pd.DataFrame({
    'Year':[2018,2020,2017,2022,2019,2021],
    'Mileage':[40000,20000,60000,10000,35000,15000],
    'Fuel':['Petrol','Diesel','Petrol','Electric','Diesel','Petrol'],
    'Price':[600000,900000,450000,1500000,700000,1100000]
})

X, y = data.drop('Price',axis=1), data['Price']
prep = ColumnTransformer([('fuel',OneHotEncoder(handle_unknown='ignore'),['Fuel'])],
                         remainder='passthrough')
model = make_pipeline(prep,RandomForestRegressor(random_state=42)).fit(X,y)

print("Predicted Price:", model.predict(pd.DataFrame({
    'Year':[2021],'Mileage':[25000],'Fuel':['Petrol']}))[0])
```

```python

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import accuracy_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC

iris = load_iris()
X_train, X_test, y_train, y_test = train_test_split(
    iris.data, iris.target, test_size=0.2, random_state=42)

models = {
    'KNN': make_pipeline(StandardScaler(), KNeighborsClassifier()),
    'Naive Bayes': GaussianNB(),
    'Decision Tree': DecisionTreeClassifier(random_state=42),
    'Logistic Regression': make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000)),
    'SVM': make_pipeline(StandardScaler(), SVC())
}

for name, model in models.items():
    model.fit(X_train, y_train)
    print(name, "Accuracy:", accuracy_score(y_test, model.predict(X_test)))
```

```python

import pandas as pd
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import LabelEncoder

data = pd.DataFrame({
    'Income':[25000,40000,50000,80000,30000,90000,60000,75000],
    'CreditHistory':['Poor','Good','Good','Good','Poor','Good','Poor','Good'],
    'Loan':['No','Yes','Yes','Yes','No','Yes','No','Yes']
})

credit = LabelEncoder()
loan = LabelEncoder()
data['CreditHistory'] = credit.fit_transform(data['CreditHistory'])
data['Loan'] = loan.fit_transform(data['Loan'])

X = data[['Income','CreditHistory']]
y = data['Loan']

model = GaussianNB().fit(X,y)
sample = pd.DataFrame({
    'Income':[55000],
    'CreditHistory':credit.transform(['Good'])
})

print("Loan Prediction:", loan.inverse_transform(model.predict(sample))[0])
```
